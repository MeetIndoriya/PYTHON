# Exception Handling

### 01) WAP to handle following exceptions:
1. ZeroDivisionError
2. ValueError
3. TypeError
#### Note: handle them using separate except blocks and also using single except block too.

```python
try:
    div = 10//0   # Divison With Zero
    print(div)
except ZeroDivisionError:
    print("ZeroDivisionError Occur")

try:
    a = int("abc")   # Value not Conveted In Valid DataType
except ValueError:
    print("Value Error Occur")

try:
    b = "10" + 5
except TypeError:    # Invalid DataType Operation
    print("TypeError Occur")
```

### 06) WAP to create an udf divide(a,b) that handles ZeroDivisionError.

```python
def divided(a,b):
    try:    
        c = a/b
    except Exception as err:
        # Standard Format ClassName: Error Message 
        print(type(err).__name__ + ": " + str(err))
    else:
        print(c)
    finally:
        print("Operation Completed Successfully")
divided(10,2)
```

### 07) WAP to accept item prices and calculate total. Raise ValueError if price is negative, otherwise print the total bill amount.

```python
class NegativeNumberError(Exception):
    pass
    
n = int(input("Enter The Item Number: "))
total = 0
try:
    for i in range(1,n+1,1):
        num = int(input("Enter The Number"))
        if (num >= 0):
            total += num
        else:
            raise NegativeNumberError("Negative Number Error")
except NegativeNumberError as err:
    print(err)
else:
    print("Total Bill Is: " + str(total))
```

### 08) Create a short program that prompts the user for a list of grades separated by commas. 
#### Split the string into individual grades and use a list comprehension to convert each string to an integer. x
#### You should use a try statement to inform the user when the values they entered cannot be converted.

```python
grade = input("Enter grades: ")

try:
    grades = [int(grade) for grade in grade.split(",")]
    print("Grades list:", grades)

except ValueError:
    print("Error: Please enter only numbers separated by commas.")
```

### 09) Accept 5 subject marks (0-100). Use assert for validation.

#### Sample Input (Error Case)
#####  Enter marks: 110 

#### Sample Output (Error Case)
##### AssertionError: Marks should be between 0 and 100.

#### Sample Input (Normal Case)
##### Valid marks entered

#### Sample Output (Normal Case)
##### Average Marks displayed

```python
total = 0
try:
    for i in range(1, 6):
        marks = int(input("Enter marks: ")

        # assert is keyword if condition returns False, AssertionError is raised:           
        assert 0 <= marks <= 100, "Marks should be between 0 and 100."
        total += marks
    print("Valid marks entered")
    average = total / 5
    print("Average Of Marks:", average)

except AssertionError as e:
    print("AssertionError:", e)
```

### 10) WAP that gets password from user. Password must have at least 8 characters and one digit. 
### Raise WeakPasswordError if invalid, print the messsage "Correct" otherwise.

```python
class WeakPasswordException(Exception):   
    pass

try:
    password = input("Enter password: ")
    if len(password) < 8 or not any(ch.isdigit() for ch in password):   # Check Password contain least 8 character and 1 digit
        raise WeakPasswordException("Password must have at least 8 characters and one digit.")

    print("Correct")

except WeakPasswordException as e:
    print("WeakPasswordError:", e)
```

### 11) WAP to raise your custom Exception named NegativeNumberError with the error message : "Cannot calculate the square root of a negative number" :
#### if the given number is negative.
#### otherwise print the square root of the given number.

```python
import math

try:
    num = float(input("Enter a number: "))
    if num < 0:
        raise NegativeNumberError("square root of a negative number")  # class Of NegativeNumber Error Made in Program 7.
    print("Square root:", math.sqrt(num))
except NegativeNumberError as err:
    print(type(err).__name__ + ": " + str(err))
```
