### 15) WAP to check whether the given number is an Armstrong Number or not.

```python
armstronNum = int(input("Enter a number: "))
n = armstronNum
power = len(str(armstronNum))
total = 0

while n > 0:
    digit = n % 10
    total += digit ** power
    n //= 10

if total == armstronNum:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

### 16) WAP to print all the perfect numbers between 1 to n.

```python
NumInput = int(input("Enter the upper limit (n): "))

print(f"Perfect numbers between 1 and {NumInput} are:")

for num in range(1, NumInput + 1):
    divisorsSum = 0
    for i in range(1, num):
        if num % i == 0:
            divisorsSum += i

    if divisorsSum == num and num != 0:
        print(num)

```
