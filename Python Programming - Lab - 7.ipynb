### 03) WAP to find tuples which have all elements divisible by K from a list of tuples.
**Sample Input:**
```
tuple_list = [(2, 4, 6), (3, 6, 9), (4, 8, 12), (5, 10, 15)]
K = 2

```
**Sample Output:**
```
Tuples divisible by 2 : [(2, 4, 6), (4, 8, 12)]
```

````python
T3 = [(2, 4, 6), (3, 6, 9), (4, 8, 12), (5, 10, 15)]

print("The original tuple list is : " + str(T3))
K = 2

l1 = []
for i in T3:
    count=0
    for j in i:
        if(j%K==0):
            count+=1
    if(count==len(i)):
        l1.append(i)
print("Tuples divisible by 2 :",str(l1))
````

### 14) WAP to create (index, value) pairs for prime numbers.

**Sample Input:**
```
T = (4, 7, 9, 11, 15)
```
**Sample Output:**
```
((1, 7), (3, 11))
```

````python
T14 = (4, 7, 9, 11, 15)

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

result14 = tuple((i, v) for i, v in enumerate(T14) if is_prime(v))
print(result14)
````

### 16) WAP to compute column-wise sum from a tuple of lists.

**Sample Input:**
```
T = ([1, 2, 3], [4, 5, 6], [7, 8, 9])
```

**Sample Output:**
```
12 15 18

```

````python
T16 = ([1, 2, 3], [4, 5, 6], [7, 8, 9])

newT16 = []
for i in range(len(T16[0])):
    s = 0
    for j in T16:
        s += j[i]
    newT16.append(s)

print(tuple(newT16))
````
