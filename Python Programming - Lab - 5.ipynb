### 1) Character Category Counter
**Problem:** Count uppercase, lowercase, digits, and special characters in a string.

**Sample Inputs & Outputs:**
- Input: `Python3@Lab` → Uppercase: 2, Lowercase: 7, Digits: 1, Special: 1
- Input: `HELLO123` → Uppercase: 5, Lowercase: 0, Digits: 3, Special: 0
- Input: `hi@#9` → Uppercase: 0, Lowercase: 2, Digits: 1, Special: 2

```python
P1str = "Python3@Lab"
P1str2 = "HELLO123"
P1str3 = "hi@#9"
lowerCounter = 0
upperCounter = 0
digitCounter = 0
SpecialCounter = 0

for x in P1str:
    if x.islower():
        lowerCounter+=1
    elif x.isupper():
        upperCounter+=1
    elif x.isdigit:
        digitCounter+=1
    else:
        SpecialCounter+=1
        
print(f"{P1str} -> Uppercase : {upperCounter} lowercase : {lowerCounter} digits : {digitCounter} Special : {SpecialCounter}")


lowerCounter = 0
upperCounter = 0
digitCounter = 0
SpecialCounter = 0

for x in P1str2:
    if x.islower():
        lowerCounter+=1
    elif x.isupper():
        upperCounter+=1
    elif x.isdigit:
        digitCounter+=1
    else:
        SpecialCounter+=1
        
print(f"{P1str2} -> Uppercase : {upperCounter} lowercase : {lowerCounter} digits : {digitCounter} Special : {SpecialCounter}")


lowerCounter = 0
upperCounter = 0
digitCounter = 0
SpecialCounter = 0

for x in P1str3:
    if x.islower():
        lowerCounter+=1
    elif x.isupper():
        upperCounter+=1
    elif x.isdigit:
        digitCounter+=1
    else:
        SpecialCounter+=1
        
print(f"{P1str3} -> Uppercase : {upperCounter} lowercase : {lowerCounter} digits : {digitCounter} Special : {SpecialCounter}")
```

### 9) String Compression
- `aaabbcccc` → `a3b2c4`
- `xxxyyyz` → `x3y3z1`
- `abcd` → `a1b1c1d1`

```python
P9str = "aaabbccc"

result = ""
count = 1

for i in range(len(P9str)):
    if i + 1 < len(P9str) and P9str[i] == P9str[i + 1]:
        count += 1
    else:
        result += P9str[i] + str(count)
        count = 1

print(P9str, "-> " , result)
```
