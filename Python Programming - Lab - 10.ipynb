### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True


count = 0
num = 2
primes = []
while count < 100:
    if is_prime(num):
        primes.append(num)
        count += 1
    num += 1

with open("primenumbers.txt", "w") as f:
    for p in primes:
        f.write(str(p) + "\n")

print("First 100 prime numbers written to primenumbers.txt")
```
